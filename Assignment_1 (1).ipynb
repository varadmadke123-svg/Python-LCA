```python
#Performing operations
a=float(input("Enter number: "))
b=float(input("Enter number: "))
c=input("Enter operation to perform: +,-,*,/ ")
if c=="+":
    print(a+b)
elif c=="-":
    print(a-b)
elif c=="*":
    print(a*b)
elif c=="/":
    if b!=0:
        print(a/b)
    else:
        print("Div by zero not allowed")
else:
    print("no operation done")
```

```python
#Creating List
print("--Creating List--")
list1=['apple','banana','cherry']
print("The list items are: ",list1)
#Access the list
print("--Access the list--")
print(len(list1))
#Change list
print("--Change list--")
list1 = ['apple', 'banana', 'cherry']
list1[1] = "blackcurrant"
print(list1)
#Add list
print("--Add List--")
list1=['apple','banana','cherry']
list1.append('orange')
print(list1)
#Remove list
print("--Remove List--")
list1=['apple','banana','cherry']
list1.remove('cherry')
print(list1)
#Sort list ascending
print("--Sort List Ascending--")
list1=['orange','mango','kiwi','pineapple','banana']
list1.sort()
print(list1)
#Sort List descending
print("--Sort List descending--")
list1=['orange','mango','kiwi','pineapple','banana']
list1.sort(reverse=True)
print(list1)
#Copy list
print("--Copy list--")
list1=['apple','banana','cherry']
list2=list1.copy()
print(list1)
#Join list
print("--Join list--")
list1=['apple','banana','cherry']
list2=['guava','papaya','orange']
list3=list1+list2
print(list3)

```

```python
#Creating Tuple
print("--Creating Tuple--")
thistuple = ("apple", "banana", "cherry")
print(thistuple)
#Accessing the tuple
print("--Accessing the tuple--")
thistuple = ("apple", "banana", "cherry")
print(thistuple[1])
#Update tuple
print("--Updating tuple--")
x = ("apple", "banana", "cherry")
y = list(x)
y[1] = "kiwi"
x = tuple(y)
print(x)
#Unpacking Tuple
print("--Unpacking Tuple--")
fruits = ("apple", "banana", "cherry")
(green, yellow, red) = fruits
print(green)
print(yellow)
print(red)
#Join Tuple
print("--Join Tuple--")
tuple1 = ("a", "b" , "c")
tuple2 = (1, 2, 3)
tuple3 = tuple1 + tuple2
print(tuple3)
#Join multiplied tuples
print("--Join multiplied tuples--")
fruits = ("apple", "banana", "cherry")
mytuple = fruits * 2
print(mytuple)
```

```python
courses=("Python","Data Science","Cloud Computing","Cyber Security","AI")
#Show 2nd course
print("The 2nd course : ",courses[1])

#Show last 2 courses
new=courses[3:5]
print("The last two courses : ",new)

#Checking inside tuple
i="Python"
if i in courses:
    print("Checking whether python is present : Yes")
else:
    print("Checking whether python is present : No")
    
#Finding the index
print("The index is :",courses.index("Cloud Computing"))

#Find the total number of courses
print("Total number of courses:", len(courses))

#Add two new course
courses = courses + ("Machine Learning", "Web Development")
print("After adding new courses:", courses)

#Drop "Cyber Security"
courses = tuple(course for course in courses if course != "Cyber Security")
print("After dropping Cyber Security:", courses)

#Arranging the updated courses alphabetically
sorted_courses = tuple(sorted(courses))
print("Sorted courses:", sorted_courses)

#Create a backup copy of sorted_courses
backup_courses = sorted_courses
print("Backup courses:", backup_courses)

#Courses selected by the student's friend
friend_courses = ("AI", "Blockchain", "Python")

#Combine the student's courses with the friend's courses
combined_courses = courses + friend_courses

# Display the final results
print("\n--- Final Results ---")
print("Final course tuple:", courses)
print("Sorted tuple:", sorted_courses)
print("Backup tuple:", backup_courses)
print("Combined tuple:", combined_courses)
```
